```python
from pathlib import Path

import cv2
import numpy as np
import pims
from PIL import Image
from tqdm import tqdm
```

```python
PATH_TO_CINE = Path('/mnt/c/Users/egorn/Desktop/WDP/fframe_dataset/cines/12-17-contact-22.cine') # Нужно указать путь до видоса

if not PATH_TO_CINE.exists():
    raise FileNotFoundError('Видоса нет')

OUT_PATH = Path('/mnt/c/Users/egorn/Desktop/WDP/processed_videos') / PATH_TO_CINE.stem # Куда сохранять картинки
OUT_PATH_NORM_NPY = OUT_PATH / 'norm_npy'
OUT_PATH_NORM_NPY.mkdir(exist_ok=True, parents=True)
```

```python
cine_images = pims.open(PATH_TO_CINE.as_posix())
```

```python
# Можно использовать индекс, чтобы посмотреть кадры в видосе и прицелиться
cine_images[23]
```

```python
START_FRAME = 0
STEP = 1 # каждый i-й кадр, 1 - все
END_FRAME = len(cine_images) # можно указать номер, -1 значит до последнего
```

```python
def normalize_cine(arr: np.ndarray) -> np.ndarray:
    ptp = arr.max(axis=(1,2)) - arr.min(axis=(1,2))
    # Handle edge case of a flat image.

    scaled_arr = (arr - arr.min(axis=(1,2)).reshape(-1, 1, 1)) / ptp.reshape(-1, 1, 1)

    return scaled_arr
```

```python
for i in tqdm(range(START_FRAME, END_FRAME, STEP), total=(END_FRAME - START_FRAME) // STEP):
    # uint8_img = cv2.normalize(np.asarray(cine_images[i]), None, 0, 255, cv2.NORM_MINMAX, dtype=cv2.CV_8U)
    norm_cine = np.asarray(normalize_cine(cine_images[i][np.newaxis]))
    np.save(OUT_PATH_NORM_NPY / f'{PATH_TO_CINE.stem}_frame_{i}.npy', norm_cine.astype(np.float32))

print(f'Обработано, данные лежат тут: {OUT_PATH_NORM_NPY}')
```
